## suma producto precio y cantidades despachadas. Despues sumar por batch y promediar expenditure totoal por batch

```python
import ast
import numpy as np
import pandas as pd

def calcular_promedio_por_h(panel, col_cant='cantidades_despachadas', col_precio='precio24', col_h='h'):
    
    def procesar_fila(row):
        # Convertir string a lista
        cantidades = np.array(ast.literal_eval(row[col_cant]), dtype=float)
        precios = np.array(ast.literal_eval(row[col_precio]), dtype=float)
        
        # Multiplicación elemento a elemento
        producto = cantidades * precios
        
        # Suma total de esa fila
        return producto.sum()
    
    # Calcular suma fila por fila
    panel['suma_producto'] = panel.apply(procesar_fila, axis=1)
    
    # Sumar por hora
    suma_por_h = panel.groupby(col_h)['suma_producto'].sum()
    
    # Promedio por hora
    promedio = suma_por_h.mean()
    
    return suma_por_h, promedio
```

```python
#suma_menor, prom_menor = calcular_promedio_por_h(panel5_menor_uni)
suma_maximo, prom_maximo = calcular_promedio_por_h(panel5_maximo_uni)
#suma_median, prom_median = calcular_promedio_por_h(panel5_median_uni)
suma_promedio, prom_promedio = calcular_promedio_por_h(panel5_promedio_uni)
```

```python
df_final = pd.DataFrame({
    #'prom_menor': [prom_menor],
    'prom_maximo': [prom_maximo],
    #'prom_median': [prom_median],
    'prom_promedio': [prom_promedio]
})

df_final/1000
# en mil millones
```

## sumar producto de cantidades y precio, hallar expnditure promedio para cada planta sobre batches.  Despues sumar los promedios

```python
def calcular_promedio_por_id(panel, col_cant='cantidades_despachadas', col_precio='precio24', col_planta='id_planta'):
    
    def procesar_fila(row):
        # Convertir string a lista
        cantidades = np.array(ast.literal_eval(row[col_cant]), dtype=float)
        precios = np.array(ast.literal_eval(row[col_precio]), dtype=float)
        
        # Multiplicación elemento a elemento
        producto = cantidades * precios
        
        # Suma total de esa fila
        return producto.sum()
    
    # Calcular suma fila por fila
    panel['suma_producto'] = panel.apply(procesar_fila, axis=1)
    
    # Sumar por hora
    suma_por_h = panel.groupby(col_planta)['suma_producto'].mean()
    
    # Promedio por hora
    promedio = suma_por_h.sum()
    
    return suma_por_h, promedio
```

```python
#suma_menor, prom_menor = calcular_promedio_por_id(panel5_menor_uni)
suma_maximo, prom_maximo = calcular_promedio_por_id(panel5_maximo_uni)
#suma_median, prom_median = calcular_promedio_por_id(panel5_median_uni)
suma_promedio, prom_promedio = calcular_promedio_por_id(panel5_promedio_uni)
```

```python
df_final = pd.DataFrame({
    #'prom_menor': [prom_menor],
    'prom_maximo': [prom_maximo],
    #'prom_median': [prom_median],
    'prom_promedio': [prom_promedio]
})

df_final/1000
# en mil millones
```
