```python
import copy, math, os, pickle, random
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

# ============================================================
# 0. SEED
# ============================================================
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

# ============================================================
# 1. SAMPLER
#    4 players, 3 demand units, 2 supply units each.
# ============================================================
class ProcurementSampler:
    def __init__(self, scenario_name):
        self.scenario_name = scenario_name
        self.n_players    = 4
        self.demand_units = 3
        self.supply_cap   = 2

        if scenario_name == "pab_should_win":
            self.cost_low, self.cost_high = 0.0, 1.0
        elif scenario_name == "uniform_should_win":
            self.cost_low, self.cost_high = 0.4, 0.6
        else:
            raise ValueError(f"Unknown scenario: {scenario_name}")

    def sample(self, batch_size):
        c = (
            torch.rand(batch_size, self.n_players, 1)
            * (self.cost_high - self.cost_low)
            + self.cost_low
        )
        return c, c.clone()

# ============================================================
# 2. BNE ANALYTICAL FORMULA
#    beta(phi) = 3*(phi+1)^2 / (4*(phi+2))
#    Valid for the PAB equilibrium on cost support [0,1].
# ============================================================
def bne_analytical(phi):
    return 3 * (phi + 1) ** 2 / (4 * (phi + 2))

# ============================================================
# 3. BIDDER NET  (bid = cost + nonnegative markup)
# ============================================================
class BidderNet(nn.Module):
    def __init__(self, hidden=10):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Linear(1, hidden),
            nn.SELU(),
            nn.Linear(hidden, hidden),
            nn.SELU(),
            nn.Linear(hidden, 1),
        )

    def forward(self, x):
        return x + torch.relu(self.layers(x))

# ============================================================
# 4. PRETRAIN
# ============================================================
def pretrain_truthful(policies, sampler, device, steps=200, batch_size=4096, lr=1e-2):
    optims = [torch.optim.SGD(p.parameters(), lr=lr) for p in policies]
    print("\n>>> Pretraining truthful bidding (bid = cost)...")
    for step in range(steps):
        c, o = sampler.sample(batch_size)
        c, o = c.to(device), o.to(device)
        loss = 0.0
        for i, pol in enumerate(policies):
            optims[i].zero_grad()
            l = ((pol(o[:, i]) - c[:, i]) ** 2).mean()
            l.backward()
            optims[i].step()
            loss += l.item()

        if step % 50 == 0 or step == steps - 1:
            print(f"  step {step:4d}/{steps} | loss={loss/len(policies):.8f}")
```

```python
# ============================================================
# 5. MULTI-UNIT CLEARING
# ============================================================
def procurement_outcome(bids, costs, demand_units, supply_cap, pricing):
    """
    bids, costs: (B, N)

    Returns
    -------
    accepted_mask : (B, N) bool
    payments      : (B, N)
    profits       : (B, N)
    buyer_cost    : (B,)
    clearing_bid  : (B,)
    units_sold    : (B, N)
    """
    B, N = bids.shape
    device = bids.device

    # Sort by bid ascending
    sorted_bids, sort_idx = bids.sort(dim=1)

    # Greedy unit allocation
    remaining = torch.full((B,), float(demand_units), device=device)
    units_sorted = torch.zeros(B, N, device=device)
    for j in range(N):
        give = torch.clamp(remaining, 0.0, float(supply_cap))
        units_sorted[:, j] = give
        remaining = remaining - give

    # Map back to original order
    units_sold = torch.zeros(B, N, device=device)
    units_sold.scatter_(1, sort_idx, units_sorted)
    accepted_mask = units_sold > 0

    # Clearing price = highest accepted bid
    LARGE = 1e9
    masked = bids.clone()
    masked[~accepted_mask] = -LARGE
    clearing_bid = masked.max(dim=1).values

    if pricing == "uniform":
        pay_per_unit = clearing_bid.unsqueeze(1).expand(B, N)
    else:
        pay_per_unit = bids

    payments = pay_per_unit * units_sold
    profits = payments - costs * units_sold
    buyer_cost = payments.sum(dim=1)

    return accepted_mask, payments, profits, buyer_cost, clearing_bid, units_sold

# ============================================================
# 6. EVALUATION
#    fixed_batch reduces noise in the reported learning curves.
# ============================================================
@torch.no_grad()
def evaluate_policies(
    policies,
    sampler,
    device,
    pricing,
    eval_batch_size=8192,
    fixed_batch=None,
):
    N = sampler.n_players

    if fixed_batch is None:
        c, o = sampler.sample(eval_batch_size)
        c, o = c.to(device), o.to(device)
    else:
        c, o = fixed_batch

    bids = torch.cat([policies[i](o[:, i]) for i in range(N)], dim=1)
    c2 = c.squeeze(-1)

    accepted_mask, payments, profits, buyer_cost, clearing, units_sold = procurement_outcome(
        bids, c2, sampler.demand_units, sampler.supply_cap, pricing
    )

    acc_f = accepted_mask.float()
    n_acc = acc_f.sum().clamp(min=1)

    grid_c = torch.linspace(
        sampler.cost_low, sampler.cost_high, 200, device=device
    ).unsqueeze(1)
    curves = [policies[i](grid_c).squeeze().cpu().numpy() for i in range(N)]
    curves = np.array(curves)

    return {
        "avg_bid": bids.mean().item(),
        "avg_markup": (bids - c2).mean().item(),
        "avg_accepted_bid": (bids * acc_f).sum().item() / n_acc.item(),
        "avg_payment": payments.sum().item() / n_acc.item(),
        "buyer_cost": buyer_cost.mean().item(),
        "supplier_profit": profits.sum(dim=1).mean().item(),
        "clearing_bid": clearing.mean().item(),
        "curves_all": curves,
        "curve_mean": curves.mean(axis=0),
        "grid": grid_c.squeeze().cpu().numpy(),
    }

# ============================================================
# 7. NPGA PSEUDOGRADIENT
# ============================================================
def flatten_params(model):
    return torch.cat([p.data.flatten() for p in model.parameters()])

def npga_gradient(player_idx, policies, sampler, device, pricing, batch_size, pop, sigma):
    pol = policies[player_idx]
    params = list(pol.parameters())
    flat = flatten_params(pol)
    d = flat.numel()
    N = sampler.n_players

    c, o = sampler.sample(batch_size)
    c, o = c.to(device), o.to(device)
    c2 = c.squeeze(-1)

    with torch.no_grad():
        bids_b = torch.cat([policies[i](o[:, i]) for i in range(N)], dim=1)
        _, _, prof_b, _, _, _ = procurement_outcome(
            bids_b, c2, sampler.demand_units, sampler.supply_cap, pricing
        )
        baseline = prof_b[:, player_idx].mean().item()

    eps = torch.randn(pop, d, device=device) * sigma
    fits = torch.zeros(pop, device=device)

    for p_idx in range(pop):
        off = 0
        for param in params:
            n = param.numel()
            param.data.add_(eps[p_idx, off:off + n].view_as(param))
            off += n

        with torch.no_grad():
            bids_p = torch.cat([policies[i](o[:, i]) for i in range(N)], dim=1)
            _, _, prof_p, _, _, _ = procurement_outcome(
                bids_p, c2, sampler.demand_units, sampler.supply_cap, pricing
            )
            fits[p_idx] = prof_p[:, player_idx].mean() - baseline

        off = 0
        for param in params:
            n = param.numel()
            param.data.sub_(eps[p_idx, off:off + n].view_as(param))
            off += n

    grad_flat = (fits.unsqueeze(1) * eps).sum(0) / (sigma**2 * pop)

    grads, off = [], 0
    for param in params:
        n = param.numel()
        grads.append(grad_flat[off:off + n].view_as(param))
        off += n

    return grads

# ============================================================
# 8. TRAINING LOOP
#    Improvements:
#      - sigma_t decreases over epochs
#      - learning rate decreases over epochs
#      - fixed evaluation batch reduces visual noise
# ============================================================
def train_npga(
    scenario_name,
    pricing,
    n_epochs=700,
    pretrain_iters=250,
    batch_size=4096,
    pop=64,
    lr=2e-3,
    sigma=2e-2,
    hidden=10,
    eval_frequency=25,
    n_runs=5,
    eval_batch_size=32768,
    seed_base=42,
    sigma_decay=0.995,
    sigma_floor_ratio=0.10,
    lr_decay=0.997,
):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"\n>>> Device: {device} | Scenario: {scenario_name} | Pricing: {pricing}")

    all_results = []
    for run in range(n_runs):
        set_seed(seed_base + run * 100)
        sampler = ProcurementSampler(scenario_name)
        N = sampler.n_players

        print(f"\n{'='*70}")
        print(
            f"RUN {run+1}/{n_runs} | N={N} | demand={sampler.demand_units} | "
            f"cap={sampler.supply_cap} | costs~U[{sampler.cost_low},{sampler.cost_high}]"
        )
        print(f"{'='*70}")

        policies = [BidderNet(hidden).to(device) for _ in range(N)]
        optims = [torch.optim.Adam(p.parameters(), lr=lr) for p in policies]
        schedulers = [
            torch.optim.lr_scheduler.ExponentialLR(opt, gamma=lr_decay)
            for opt in optims
        ]

        pretrain_truthful(
            policies,
            sampler,
            device,
            steps=pretrain_iters,
            batch_size=batch_size,
        )

        eval_c, eval_o = sampler.sample(eval_batch_size)
        eval_c, eval_o = eval_c.to(device), eval_o.to(device)
        fixed_eval = (eval_c, eval_o)

        res = {
            k: []
            for k in [
                "epoch",
                "buyer_cost",
                "supplier_profit",
                "avg_bid",
                "avg_markup",
                "avg_accepted_bid",
                "avg_payment",
                "clearing_bid",
                "grad_norm",
                "curve_mean",
                "curves_all",
                "grid",
                "sigma_t",
                "lr_t",
            ]
        }

        sigma0 = sigma
        sigma_min = sigma * sigma_floor_ratio

        for epoch in range(n_epochs):
            sigma_t = max(sigma_min, sigma0 * (sigma_decay ** epoch))
            grad_norms = []

            for i in range(N):
                grads = npga_gradient(
                    i,
                    policies,
                    sampler,
                    device,
                    pricing,
                    batch_size,
                    pop,
                    sigma_t,
                )
                optims[i].zero_grad()
                for param, g in zip(policies[i].parameters(), grads):
                    param.grad = -g
                nn.utils.clip_grad_norm_(policies[i].parameters(), 1.0)
                optims[i].step()
                grad_norms.append(torch.cat([g.view(-1) for g in grads]).norm().item())

            for sch in schedulers:
                sch.step()

            if epoch % eval_frequency == 0 or epoch == n_epochs - 1:
                ev = evaluate_policies(
                    policies,
                    sampler,
                    device,
                    pricing,
                    eval_batch_size=eval_batch_size,
                    fixed_batch=fixed_eval,
                )

                res["epoch"].append(epoch)
                for k in [
                    "buyer_cost",
                    "supplier_profit",
                    "avg_bid",
                    "avg_markup",
                    "avg_accepted_bid",
                    "avg_payment",
                    "clearing_bid",
                ]:
                    res[k].append(ev[k])

                res["grad_norm"].append(float(np.mean(grad_norms)))
                res["curve_mean"].append(ev["curve_mean"])
                res["curves_all"].append(ev["curves_all"])
                res["grid"].append(ev["grid"])
                res["sigma_t"].append(float(sigma_t))
                res["lr_t"].append(float(optims[0].param_groups[0]["lr"]))

                print(
                    f"[{epoch:4d}] BuyerCost={ev['buyer_cost']:.4f} | "
                    f"AvgBid={ev['avg_bid']:.4f} | "
                    f"Clear={ev['clearing_bid']:.4f} | "
                    f"Markup={ev['avg_markup']:.4f} | "
                    f"Sigma={sigma_t:.5f} | "
                    f"LR={optims[0].param_groups[0]['lr']:.5e} | "
                    f"GradNorm={np.mean(grad_norms):.2e}"
                )

        all_results.append(res)

    print("\n>>> Training finished.")
    return all_results

# ============================================================
# 9. EXPORT HELPERS
# ============================================================
def final_metric_mean(results_list, metric):
    vals = [float(r[metric][-1]) for r in results_list]
    return float(np.mean(vals))

def build_summary_table(results_dict):
    rows = []
    for name, res in results_dict.items():
        rows.append(
            {
                "result_name": name,
                "buyer_cost_final_mean": final_metric_mean(res, "buyer_cost"),
                "avg_bid_final_mean": final_metric_mean(res, "avg_bid"),
                "avg_markup_final_mean": final_metric_mean(res, "avg_markup"),
                "clearing_bid_final_mean": final_metric_mean(res, "clearing_bid"),
                "supplier_profit_final_mean": final_metric_mean(res, "supplier_profit"),
                "n_runs": len(res),
            }
        )
    return pd.DataFrame(rows)

def export_results_bundle(results_dict, export_dir="simulationPABvsPAC_vfinal/exports"):
    export_path = Path(export_dir)
    export_path.mkdir(parents=True, exist_ok=True)

    for name, obj in results_dict.items():
        with open(export_path / f"{name}.pkl", "wb") as f:
            pickle.dump(obj, f)

    with open(export_path / "all_results.pkl", "wb") as f:
        pickle.dump(results_dict, f)

    summary = build_summary_table(results_dict)
    summary.to_csv(export_path / "final_summary.csv", index=False)

    print(f"\nSaved exports in: {export_path.resolve()}")
    for name in results_dict:
        print(f"  - {name}.pkl")
    print("  - all_results.pkl")
    print("  - final_summary.csv")

    return summary
```

```python
# ============================================================
# 10. COMPARISON
# ============================================================
def mean_band(results_list, metric, band="ci95"):
    arr = np.array([r[metric] for r in results_list], dtype=float)
    mean = arr.mean(axis=0)
    sd = arr.std(axis=0, ddof=1) if arr.shape[0] > 1 else np.zeros_like(mean)

    if band == "sd":
        lo = mean - sd
        hi = mean + sd
    elif band == "ci95":
        half = 1.96 * sd / np.sqrt(arr.shape[0])
        lo = mean - half
        hi = mean + half
    elif band == "iqr":
        lo = np.quantile(arr, 0.25, axis=0)
        hi = np.quantile(arr, 0.75, axis=0)
    else:
        raise ValueError("band must be 'sd', 'ci95', or 'iqr'")

    return mean, lo, hi

def compare_pricing(res_u, res_p, scenario_name):
    cu = np.mean([r["buyer_cost"][-1] for r in res_u])
    cp = np.mean([r["buyer_cost"][-1] for r in res_p])
    winner = "uniform" if cu < cp else "pay_as_bid"

    print(f"\n{'#'*70}")
    print(f"SCENARIO: {scenario_name}")
    print(f"Buyer cost | uniform    = {cu:.6f}")
    print(f"Buyer cost | pay_as_bid = {cp:.6f}")
    print(f"WINNER: {winner}")
    print(f"{'#'*70}\n")
    return winner

# ============================================================
# 11. PLOTTING
# ============================================================
def paper_style():
    plt.rcParams.update({
        "figure.dpi": 140,
        "savefig.dpi": 300,
        "font.size": 11,
        "axes.titlesize": 12,
        "axes.labelsize": 11,
        "legend.fontsize": 10,
        "lines.linewidth": 2.0,
        "axes.spines.top": False,
        "axes.spines.right": False,
        "legend.frameon": False,
    })

def plot_with_bne(
    res_uniform,
    res_pab,
    title_prefix,
    show_bne=True,
    save_dir=None,
    band="ci95",
    alpha_band=0.16,
):
    """
    3-panel figure:
      (a) Buyer expenditure over training
      (b) Average bid over training
      (c) Final equilibrium bid function + analytical BNE
    """
    paper_style()
    uc, pc = "steelblue", "firebrick"
    bne_c = "black"

    epochs = np.array(res_uniform[0]["epoch"])

    u_exp_m, u_exp_lo, u_exp_hi = mean_band(res_uniform, "buyer_cost", band=band)
    p_exp_m, p_exp_lo, p_exp_hi = mean_band(res_pab,     "buyer_cost", band=band)
    u_bid_m, u_bid_lo, u_bid_hi = mean_band(res_uniform, "avg_bid",    band=band)
    p_bid_m, p_bid_lo, p_bid_hi = mean_band(res_pab,     "avg_bid",    band=band)

    cu = np.array([r["curve_mean"][-1] for r in res_uniform])
    cp = np.array([r["curve_mean"][-1] for r in res_pab])
    grid = res_uniform[0]["grid"][-1]

    cu_m = cu.mean(axis=0)
    cp_m = cp.mean(axis=0)

    if band == "sd":
        cu_half = cu.std(axis=0, ddof=1) if cu.shape[0] > 1 else np.zeros_like(cu_m)
        cp_half = cp.std(axis=0, ddof=1) if cp.shape[0] > 1 else np.zeros_like(cp_m)
        cu_lo, cu_hi = cu_m - cu_half, cu_m + cu_half
        cp_lo, cp_hi = cp_m - cp_half, cp_m + cp_half
    elif band == "ci95":
        cu_sd = cu.std(axis=0, ddof=1) if cu.shape[0] > 1 else np.zeros_like(cu_m)
        cp_sd = cp.std(axis=0, ddof=1) if cp.shape[0] > 1 else np.zeros_like(cp_m)
        cu_half = 1.96 * cu_sd / np.sqrt(cu.shape[0])
        cp_half = 1.96 * cp_sd / np.sqrt(cp.shape[0])
        cu_lo, cu_hi = cu_m - cu_half, cu_m + cu_half
        cp_lo, cp_hi = cp_m - cp_half, cp_m + cp_half
    elif band == "iqr":
        cu_lo, cu_hi = np.quantile(cu, 0.25, axis=0), np.quantile(cu, 0.75, axis=0)
        cp_lo, cp_hi = np.quantile(cp, 0.25, axis=0), np.quantile(cp, 0.75, axis=0)
    else:
        raise ValueError("band must be 'sd', 'ci95', or 'iqr'")
    import matplotlib.pyplot as plt
    
    plt.rcParams["axes.grid"] = False
    
    fig, axes = plt.subplots(1, 3, figsize=(15.5, 4.4))
    
    # (a) Buyer expenditure
    ax = axes[0]
    ax.plot(epochs, u_exp_m, color=uc, label="Uniform")
    ax.fill_between(epochs, u_exp_lo, u_exp_hi, alpha=alpha_band, color=uc)
    ax.plot(epochs, p_exp_m, color=pc, label="Pay-as-bid")
    ax.fill_between(epochs, p_exp_lo, p_exp_hi, alpha=alpha_band, color=pc)
    ax.set_title(f"{title_prefix}\nBuyer expenditure")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Expected buyer cost")
    ax.grid(False, which="both")
    ax.legend(loc="best")
    
    # (b) Average bid
    ax = axes[1]
    ax.plot(epochs, u_bid_m, color=uc, label="Uniform")
    ax.fill_between(epochs, u_bid_lo, u_bid_hi, alpha=alpha_band, color=uc)
    ax.plot(epochs, p_bid_m, color=pc, label="Pay-as-bid")
    ax.fill_between(epochs, p_bid_lo, p_bid_hi, alpha=alpha_band, color=pc)
    ax.set_title("Average bid")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Bid")
    ax.grid(False, which="both")
    ax.legend(loc="best")
    
    # (c) Final bid function
    ax = axes[2]
    ax.plot(grid, cu_m, color=uc, label="Uniform (learned)")
    ax.fill_between(grid, cu_lo, cu_hi, alpha=alpha_band, color=uc)
    ax.plot(grid, cp_m, color=pc, label="Pay-as-bid (learned)")
    ax.fill_between(grid, cp_lo, cp_hi, alpha=alpha_band, color=pc)
    
    if show_bne:
        ax.plot(grid, bne_analytical(grid), "--", color=bne_c, label="Analytical BNE")
    
    ax.plot(grid, grid, ":", color="gray", lw=1.2, label="Truthful bid")
    ax.set_title("Final bid function")
    ax.set_xlabel("Cost")
    ax.set_ylabel("Bid")
    ax.grid(False, which="both")
    ax.legend(loc="best")
    
    fig.tight_layout()
    plt.show()
```

```python
# ============================================================
# 12. RUN BOTH SCENARIOS
# ============================================================
common = dict(
    n_epochs=600,
    pretrain_iters=250,
    batch_size=4096,
    pop=64,
    lr=2e-3,
    sigma=0.02,
    hidden=10,
    eval_frequency=25,
    n_runs=5,
    eval_batch_size=32768,
    sigma_decay=0.995,
    sigma_floor_ratio=0.10,
    lr_decay=0.997,
)

# Case 1: U(0,1) -> PAB expected to win
print("\n" + "="*70)
print("CASE 1: U(0,1) -> PAB expected to win")
print("="*70)
res_u1 = train_npga("pab_should_win",     "uniform",    **common)
res_p1 = train_npga("pab_should_win",     "pay_as_bid", **common)
compare_pricing(res_u1, res_p1, "pab_should_win [U(0,1)]")
plot_with_bne(
    res_u1,
    res_p1,
    title_prefix="Case 1: discriminatory pricing cheaper",
    show_bne=True,
    save_dir="simulationPABvsPAC_vfinal/figures",
    band="ci95",
)

```

```python
plot_with_bne(
    res_u1,
    res_p1,
    title_prefix="",#Case 1: discriminatory pricing cheaper
    show_bne=True,
    save_dir="simulationPABvsPAC_vfinal/figures",
    band="ci95",
)
```

```python
common = dict(
        n_epochs=600, pretrain_iters=200,
        batch_size=4096, pop=64, lr=3e-3, sigma=0.02,
        hidden=10, eval_frequency=50, n_runs=5, eval_batch_size=8192,
    )
# Case 2: U(0.4,0.6) -> Uniform expected to win
print("\n" + "="*70)
print("CASE 2: U(0.4,0.6) -> Uniform expected to win")
print("="*70)
res_u2 = train_npga("uniform_should_win", "uniform",    **common)
res_p2 = train_npga("uniform_should_win", "pay_as_bid", **common)
compare_pricing(res_u2, res_p2, "uniform_should_win [U(0.4,0.6)]")
plot_with_bne(
    res_u2,
    res_p2,
    title_prefix="Case 2: uniform pricing cheaper",
    show_bne=False,
    save_dir="simulationPABvsPAC_vfinal/figures",
    band="ci95",
)
```

```python
# ============================================================
# 13. EXPORT RESULTS
#    This saves the objects you asked for:
#      res_u1, res_p1, res_u2, res_p2
# ============================================================
results_dict = {
    "res_u1": res_u1,
    "res_p1": res_p1,
    "res_u2": res_u2,
    "res_p2": res_p2,
}

summary_df = export_results_bundle(
    results_dict,
    export_dir="simulationPABvsPAC_vfinal/exports",
)
summary_df
```
