## Merge

```python
# Merge df_merged con df_trm
df_final = pd.merge(df_merged, df_trm, on="Fecha", how="left")


df_final.columns
```

```python
df_final=df_final[['Fecha', 'ACPM', 'CARBON', 'GAS', 'GAS NI', 'COMBUSTOLEO', 'VOM', 'CERE',  'FAZNI',
    'VALOR']]
```

```python
df_final = df_final.rename(columns={
    "VALOR": "trm"
})
```

```python


# Paso 1: Limpiar TRM y convertir a float
df_final["trm"] = df_final["trm"].astype(str)  # asegurar que es string
df_final["trm"] = df_final["trm"].str.replace("$", "", regex=False)
df_final["trm"] = df_final["trm"].str.replace(".", "", regex=False)  # elimina separador de miles
df_final["trm"] = df_final["trm"].str.replace(",", ".", regex=False)  # convierte decimal
df_final["trm"] = pd.to_numeric(df_final["trm"], errors="coerce")  # convierte a float, NaN si no hay valor

# Paso 2: Rellenar valores faltantes con último valor válido
df_final["trm"] = df_final["trm"].ffill()
df_final["trm"][0]=float(4149.24)
```

## Costos de ingenieria

```python
# Fechas referencia
fechas_ref = pd.Index(sorted(pd.to_datetime(df['Fecha'].unique())))

def fill_fechas(df):
    return (df.assign(Fecha=pd.to_datetime(df['Fecha']))
              .set_index('Fecha')
              .reindex(fechas_ref)
              .ffill()
              .bfill()
              .rename_axis('Fecha')
              .reset_index())

# Aplicar a ambos
df_final = fill_fechas(df_final)
```

```python
df['Fecha']= pd.to_datetime(df['Fecha'])

df_ingenieria1=pd.merge(df[['CodigoPlanta','phi', 'precio_d', 'Fecha', 
       'id_planta', 'Combustible por Defecto',
       'Nombre Recurso', 'Agente Representante', 'available_q', 'daily_reman_demand']], df_final, on='Fecha', how='left' )
df_ingenieria1 = df_ingenieria1.sort_values('Fecha').ffill().bfill()
```

```python
df_eficiencia['Nombre'] = df_eficiencia['Nombre'].str.strip()
df_ingenieria=pd.merge(df_ingenieria1, df_eficiencia[['Agente', 'Nombre',
       'eficiencia']], left_on='Nombre Recurso', right_on='Nombre', how='left' )
df_ingenieria['Nombre Recurso'] = df_ingenieria1['Nombre Recurso'].str.strip()
# Valores CP
cp_gas = 1.037
cp_carbon = 24.00
cp_acpm = 5.825
cp_mezcla = np.mean([cp_gas, cp_carbon])

# Diccionario de equivalencias
cp_map = {
    'GAS': cp_gas,
    'CARBON': cp_carbon,
    'ACPM': cp_acpm,
    'MEZCLA GAS - CARBÓN': cp_mezcla
}

# Crear columna CP en df_ingenieria
df_ingenieria['CP'] = df_ingenieria['Combustible por Defecto'].map(cp_map)
```

```python
len(df_ingenieria["CodigoPlanta"].unique())
```

```python
def asignar_pfuel(row):
    cdef = row["Combustible por Defecto"]
    # caso MEZCLA GAS - CARBÓN
    if cdef == "MEZCLA GAS - CARBÓN":
        return min(row["GAS"], row["CARBON"])  # promedio simple
    # caso normal: solo columna correspondiente
    elif cdef in row.index:
        return row[cdef]
    else:
        return None  # si no encuentra columna
df_ingenieria["pfuel"] = df_ingenieria.apply(asignar_pfuel, axis=1)
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import os

# Asegurar que Fecha sea datetime
df_ingenieria['Fecha'] = pd.to_datetime(df_ingenieria['Fecha'])

# Carpeta de salida
output_path = "../results/graficasCostos"
os.makedirs(output_path, exist_ok=True)

for planta, df_p in df_ingenieria.groupby('CodigoPlanta'):
    
    nombre = df_p['Nombre Recurso'].iloc[0]
    combustible = df_p['Combustible por Defecto'].iloc[0]
    codigo = df_p['CodigoPlanta'].iloc[0]

    fig, ax = plt.subplots(figsize=(10,5))
    
    # Observed bid (línea principal)
    ax.plot(df_p['Fecha'], df_p['precio_d'], 
        '-', color='#2ca02c', linewidth=2.4, label='Observed bid')  # verde oscuro
    ax.plot(df_p['Fecha'], df_p['phi'], 
        '--', color='#98df8a', linewidth=1.8, label='MC_constant')
    ax.plot(df_p['Fecha'], df_p['phi_ramping'], 
        '-.', color='#1f7a1f', linewidth=1.8, label='MC_ramping')  # verde medio

    # MC_Eng
    ax.plot(df_p['Fecha'], df_p['phi_eng'], 
            ':', color='#6e6e6e', linewidth=1.8, label='MC_Eng')


    # Formato eje X
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    ax.xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.xticks(rotation=0)

    ax.set_title(f"{codigo}-{nombre} ({combustible})")
    ax.set_xlabel("Date")
    ax.set_ylabel("COP/kWh")
    ax.legend()
    #ax.grid(True, alpha=0.3)

    plt.tight_layout()

    # Guardar figura
    filename = f"{codigo}_{nombre}.png".replace(" ", "_")
    plt.savefig(os.path.join(output_path, filename), dpi=300, bbox_inches="tight")
    plt.show()
    #plt.close()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Ruta de salida
output_path = "../results/graficasCostos"
os.makedirs(output_path, exist_ok=True)

# Asegurar formato de fecha
df_ingenieria['Fecha'] = pd.to_datetime(df_ingenieria['Fecha'])

# Agrupar por día
df_time = df_ingenieria.groupby('Fecha')[["phi", "phi_eng", "precio_d", 'phi_ramping']].mean()

# Crear gráfica
plt.figure(figsize=(12,6))

plt.plot(df_time.index, df_time['precio_d'], 
         color='#2ca02c', linestyle='-', linewidth=2.4, label='Observed bid')  # verde oscuro

plt.plot(df_time.index, df_time['phi'], 
         color='#98df8a', linestyle='--', linewidth=1.8, label='MC')  # verde claro

plt.plot(df_time.index, df_time['phi_ramping'], 
         color='#1f7a1f', linestyle='-.', linewidth=1.8, label='MC_ramping')  # verde medio

plt.plot(df_time.index, df_time['phi_eng'], 
         color='#6e6e6e', linestyle=':', linewidth=1.8, label='MC_Eng')

plt.title('Estimated Marginal Costs vs Observed Bid Prices (Daily Average)')
plt.xlabel('Date')
plt.ylabel('COP/kWh')
plt.legend()
plt.grid(False)

plt.tight_layout()

# Guardar figura
file_path = os.path.join(output_path, "serie_temporal_costos.png")
plt.savefig(file_path, dpi=300, bbox_inches='tight')

# Mostrar
plt.show()
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Ruta de salida
output_path = "../results/graficasCostos"
os.makedirs(output_path, exist_ok=True)

# Asegurar formato de fecha
df_ingenieria['Fecha'] = pd.to_datetime(df_ingenieria['Fecha'])

# Agrupar y calcular estadísticas
df_stats = df_ingenieria.groupby('Fecha')[["phi", "phi_eng", "precio_d", "phi_ramping"]].agg(['mean','std','count'])

# Parámetro para IC 95%
z = 1.96

# Intervalos para phi
phi_mean = df_stats[("phi","mean")]
phi_ci_upper = phi_mean + z * df_stats[("phi","std")] / np.sqrt(df_stats[("phi","count")])
phi_ci_lower = phi_mean - z * df_stats[("phi","std")] / np.sqrt(df_stats[("phi","count")])

# (Opcional) Intervalos para phi_ramping
phi_r_mean = df_stats[("phi_ramping","mean")]
phi_r_ci_upper = phi_r_mean + z * df_stats[("phi_ramping","std")] / np.sqrt(df_stats[("phi_ramping","count")])
phi_r_ci_lower = phi_r_mean - z * df_stats[("phi_ramping","std")] / np.sqrt(df_stats[("phi_ramping","count")])

# Crear gráfica
plt.figure(figsize=(12,6))

# Observed bid
plt.plot(df_stats.index, df_stats[("precio_d","mean")], 
         color='#2ca02c', linestyle='-', linewidth=2.4, label='Observed bid')

# MC
plt.plot(df_stats.index, phi_mean, 
         color='#98df8a', linestyle='--', linewidth=1.8, label='MC')

# Banda de confianza MC
plt.fill_between(df_stats.index, phi_ci_lower, phi_ci_upper, 
                 color='#98df8a', alpha=0.25, label='CI 95% (MC)')

# MC_ramping
plt.plot(df_stats.index, phi_r_mean, 
         color='#1f7a1f', linestyle='-.', linewidth=1.8, label='MC_ramping')

# Banda de confianza MC_ramping
plt.fill_between(df_stats.index, phi_r_ci_lower, phi_r_ci_upper, 
                 color='#1f7a1f', alpha=0.2, label='CI 95% (MC_ramping)')

# MC_Eng
plt.plot(df_stats.index, df_stats[("phi_eng","mean")], 
         color='#6e6e6e', linestyle=':', linewidth=1.8, label='MC_Eng')

# Estética
plt.title('Estimated Marginal Costs vs Observed Bid Prices (Daily Average)')
plt.xlabel('Date')
plt.ylabel('COP/kWh')
plt.legend()
plt.grid(False)

plt.tight_layout()

# Guardar figura
file_path = os.path.join(output_path, "serie_temporal_costos_ci.png")
plt.savefig(file_path, dpi=300, bbox_inches='tight')

# Mostrar
plt.show()
```
